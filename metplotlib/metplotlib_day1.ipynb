```python
import matplotlib
import matplotlib.pyplot as plt 
```

```python
x=[1,2,3,4,5]
y=[10,20,30,40,5]
plt.plot(x,y)
plt.show()
```

```python

months=["jan","feb","mar","apr", "may"]
sales=[121,2132,2423,2342,234]
plt.plot(months,sales)
plt.title("monthly sales")
plt.xlabel("Months")
plt.ylabel("sales")
plt.show()
```

```python
plt.plot(
    months,
    sales,
    color="red",
    linestyle="--",
    linewidth=0.5,marker="s"
)
```

```python
months=["jan","feb","mar","apr", "may"]
sales_2025=[100,150,120,180,220]
sales_2026=[120,250,220,280,320]
plt.plot(months,sales_2025,marker='o')
plt.plot(months,sales_2026,marker='s')
plt.title("Montly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend("Center")
plt.show()
```
